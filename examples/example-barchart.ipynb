# Barchart Widget Examples

This notebook demonstrates all variations of the barchart widget:
1. Simple Barchart
2. Stacked Barchart
3. Mix Barchart
4. Positive-Negative Barchart
5. Stacked by Sign Barchart
6. Biaxial Barchart

```typescript
import { barchart, row, scatter } from 'tslab-widgets';
```

## Barchart (Simple, Stacked and Mix)

A set of barchart examples.

```typescript
const dataWithNegatives = [ { name: 'Page A', uv: 4000, pv: 2400, amt: 2400 }, { name: 'Page B', uv: -3000, pv: 1398, amt: 2210 }, { name: 'Page C', uv: -2000, pv: -9800, amt: 2290 }, { name: 'Page D', uv: 2780, pv: 3908, amt: 2000 }, { name: 'Page E', uv: -1890, pv: 4800, amt: 2181 }, { name: 'Page F', uv: 2390, pv: -3800, amt: 2500 }, { name: 'Page G', uv: 3490, pv: 4300, amt: 2100 } ];
// Simple
barchart({
  data: dataWithNegatives,
  xAxisKey: 'name',
  variation: 'simple',
  title: 'Simple Barchart',
  bars: [
    { dataKey: 'pv', fill: '#8884d8' },
    { dataKey: 'uv', fill: '#82ca9d' }
  ],
  width: 600,
});
// Stacked
barchart({
  data: dataWithNegatives,
  xAxisKey: 'name',
  variation: 'stacked',
  title: 'Stacked Barchart',
  bars: [
    { dataKey: 'pv', stackId: 'a', fill: '#8884d8' },
    { dataKey: 'uv', stackId: 'a', fill: '#82ca9d' }
  ],
  width: 600,
});
// Mix
barchart({
  data: dataWithNegatives,
  xAxisKey: 'name',
  variation: 'mix',
  title: 'Mix Barchart',
  bars: [
    { dataKey: 'pv', stackId: 'a', fill: '#8884d8' },
    { dataKey: 'amt', stackId: 'a', fill: '#82ca9d' },
    { dataKey: 'uv', fill: '#ffc658' }
  ],
  width: 600,
});

// Custom colors and no legend
barchart({
  data: dataWithNegatives,
  xAxisKey: 'name',
  variation: 'simple',
  title: 'Custom Styled Barchart',
  showLegend: false,
  showGrid: false,
  // width: 700,
  margin: { top: 20, right: 30, left: 20, bottom: 5 },
  bars: [
    {
      dataKey: 'pv',
      fill: '#ff6b6b',
      radius: [8, 8, 0, 0]
    },
    {
      dataKey: 'uv',
      fill: '#4ecdc4',
      radius: [8, 8, 0, 0]
    },
    {
      dataKey: 'amt',
      fill: '#ffe66d',
      radius: [8, 8, 0, 0]
    }
  ]
});
```

### Using row layout

Display multiple charts side by side.

```typescript
const data = [ { name: 'Page A', uv: 4000, pv: 2400, amt: 2400 }, { name: 'Page B', uv: 3000, pv: 1398, amt: 2210 }, { name: 'Page C', uv: 2000, pv: 9800, amt: 2290 }, { name: 'Page D', uv: 2780, pv: 3908, amt: 2000 }, { name: 'Page E', uv: 1890, pv: 4800, amt: 2181 }, { name: 'Page F', uv: 2390, pv: 3800, amt: 2500 }, { name: 'Page G', uv: 3490, pv: 4300, amt: 2100 } ];

row([
  {
    html: barchart.html({
      data,
      xAxisKey: 'name',
      variation: 'simple',
      title: 'Simple Barchart',
      bars: [
        {
          dataKey: 'pv',
          fill: '#8884d8',
          radius: [10, 10, 0, 0],
          activeBar: { fill: 'pink', stroke: 'blue' }
        },
        {
          dataKey: 'uv',
          fill: '#82ca9d',
          radius: [10, 10, 0, 0],
          activeBar: { fill: 'gold', stroke: 'purple' }
        }
      ],
    }),
    flex: 1,
  },
  {
    html: barchart.html({
      data,
      xAxisKey: 'name',
      variation: 'stacked',
      title: 'Stacked',
      bars: [
        { dataKey: 'pv', stackId: 'a', fill: '#8884d8' },
        { dataKey: 'uv', stackId: 'a', fill: '#82ca9d' }
      ]
    }),
    flex: 1
  }
], {});
```
