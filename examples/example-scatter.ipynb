# Scatter Chart Example

This example recreates the official AG Charts scatter plot example showing the relationship between height and weight for males and females.

```typescript
import { scatter } from 'tslab-widgets';
import { getScatterData } from './testData';


const allData = getScatterData();
scatter({
  title: "Weight vs Height",
  subtitle: "by gender",
  series: [
    {
      title: "Male",
      data: allData.maleHeightWeight,
      xKey: "height",
      xName: "Height",
      yKey: "weight",
      yName: "Weight",
      shape: "diamond",
      fill: "#f30000",
    },
    {
      title: "Female",
      data: allData.femaleHeightWeight,
      xKey: "height",
      xName: "Height",
      yKey: "weight",
      yName: "Weight",
      shape: "circle",
      fill: "#82ca9d",
    },
  ],
  xAxisTitle: "Height",
  yAxisTitle: "Weight",
  xAxisFormatter: "(params) => params.value + 'cm'",
  yAxisFormatter: "(params) => params.value + 'kg'",
  width: 700,
  height: 500,
  showLegend: true,
});
```
