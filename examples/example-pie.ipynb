# Pie Chart Examples

This notebook demonstrates various pie chart configurations using the tslab-widgets library.

```typescript
import { pie, row } from  from 'tslab-widgets';
```

## Basic Pie and Donut Charts

A simple pie chart with default colors.

```typescript
pie({
  data: [
    { name: 'Desktop', value: 400 },
    { name: 'Mobile', value: 300 },
    { name: 'Tablet', value: 200 },
    { name: 'Other', value: 100 },
  ],
  title: 'Device Distribution',
  showLabels: true,
  useCustomLabels: true,
  showLabelLines: false,
});

pie({
  data: [
    { name: 'React', value: 450, fill: '#10b9f1' },
    { name: 'Vue', value: 280, fill: '#ff0000ff' },
    { name: 'Angular', value: 180, fill: '#ff4ff6' },
    { name: 'Svelte', value: 90, fill: '#5fff50ff' },
  ],
  innerRadius: 50,
  outerRadius: 80,
  title: 'Framework Usage',
  showLabels: true,
});
```

## Two-Level Pie Chart with Row Layout

A complex chart with both inner and outer data layers with a row layout.

```typescript
const data01 = [ { name: 'Group A', value: 400 }, { name: 'Group B', value: 300 }, { name: 'Group C', value: 300 }, { name: 'Group D', value: 200 } ];
const data02 = [ { name: 'A1', value: 100 }, { name: 'A2', value: 300 }, { name: 'B1', value: 100 }, { name: 'B2', value: 80 }, { name: 'B3', value: 40 }, { name: 'B4', value: 30 }, { name: 'B5', value: 50 }, { name: 'C1', value: 100 }, { name: 'C2', value: 200 }, { name: 'D1', value: 150 }, { name: 'D2', value: 50 } ];

row([
  {
    html: pie.html({
      data: data01,
      innerData: data02,
      title: 'Two-Level Pie Chart',
      showLabels: true,
    }),

  },
  {
    html: pie.html({
      data: [
        { name: 'Q1', value: 250 },
        { name: 'Q2', value: 320 },
        { name: 'Q3', value: 280 },
        { name: 'Q4', value: 350 },
      ],
      title: 'Quarterly Revenue',
      animate: false,
      showLabels: true,
    }),
  }
]);
```
