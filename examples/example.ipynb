## TsLab Widgets

A set of widgets for ts-lab environment including financial charts, CSV tables, gauges, and JSON viewers.

### Chart Syncing

Charts with the same `syncKey` will automatically sync in real-time:
- **Crosshair position**: Move your mouse over any chart to see the crosshair mirrored on all synced charts
- **Time scale (zoom/pan)**: Zoom or pan on any chart and all synced charts will follow

All chart types are supported (Candlestick, Line, Area, Bar, Baseline, Histogram). No button clicking required!

## Gauge, Table and JSON components

Multiple widgets ready for data visualization.

```typescript
gauge({ title: 'General Indicator', value: 65, size: 'S', unit: '%' });

table({
  options: {
    rowData: [
      { make: "Tesla", model: "Model Y", price: 64950, electric: true },
      { make: "Ford", model: "F-Series", price: 33850, electric: false },
      { make: "Toyota", model: "Corolla", price: 29600, electric: false },
    ]
  },
  height: 150,
  width: 800,
});

json({ data: { some: 'json data', testArray: [1,2,3,4] }});
```

## Multiple Cells Layout

Create a dashboard with multiple widgets in a row.

```typescript
// Gauge sections customization
const sections = [{ value: 25, fill: "#f5420bff" }, { value: 25, fill: "#0b80f5ff" }];

// First row shows 4 gauges
row(
  [
    { html: gauge.html({ title: 'General', value: 58, size: 'L', unit: '%', sections }) },
    { html: gauge.html({ title: 'CPU Usage', value: 65, size: 'S', unit: '%', sections }) },
    { html: gauge.html({ title: 'Memory', value: 28, size: 'S', unit: '%', sections }) },
    { html: gauge.html({ title: 'Disk I/O', value: 82, size: 'S', unit: '%', sections }) }
  ], { minHeight: "180px" }
);

// Second row shows a table and JSOn widgets
row([
  {
    html: table.html({
      options: {
        rowData: [
          { metric: 'CPU Usage', current: 65, average: 58, peak: 92 },
          { metric: 'Memory', current: 48, average: 52, peak: 78 },
          { metric: 'Disk I/O', current: 82, average: 45, peak: 95 },
        ]
      },
      height: 150,
      width: 600
    }),
    minWidth: '400px'
  },
  { html: json.html({ data: { status: 'active', count: 42 }, width: 150 }) }
]);
```
